## Contoh program

**Transformer classifier kecil**  
Keras MultiHeadAttention menerima query dan value; self-attention memakai tensor yang sama.


```python
import keras
from keras import layers

inputs = keras.Input((80,), dtype="int32")
x = layers.Embedding(5000, 64, mask_zero=True)(inputs)
attn = layers.MultiHeadAttention(num_heads=4, key_dim=16)(x, x)
x = layers.LayerNormalization()(x + attn)
ffn = layers.Dense(128, activation="relu")(x)
ffn = layers.Dense(64)(ffn)
x = layers.LayerNormalization()(x + ffn)
x = layers.GlobalAveragePooling1D()(x)
x = layers.Dropout(.3)(x)
outputs = layers.Dense(3, activation="softmax")(x)
model = keras.Model(inputs, outputs)
model.compile("adam", "sparse_categorical_crossentropy", metrics=["accuracy"])
model.summary()
```
