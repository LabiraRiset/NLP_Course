```python
import pandas as pd
import re
from collections import Counter

data = pd.DataFrame({
    "text": [
        "Aplikasinya mudah dipakai!",
        "ga recommended, sering error",
        "Layanan cepat dan sangat membantu"
    ],
    "label": ["positif", "negatif", "positif"]
})

def tokens(text):
    return re.findall(r"[a-zA-ZÀ-ÿ]+", text.lower())

data["n_char"] = data.text.str.len()
data["n_token"] = data.text.map(lambda x: len(tokens(x)))
print(data)
print("Distribusi label:", data.label.value_counts().to_dict())
print("Kosakata:", Counter(t for x in data.text for t in tokens(x)))
```
