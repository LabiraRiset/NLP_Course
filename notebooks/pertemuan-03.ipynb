## Contoh program

**Baseline bebas leakage**  
Pipeline memastikan vectorizer hanya di-fit pada data train.


```python
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report

texts = ["bagus sekali", "sangat buruk", "suka aplikasinya",
         "sering gagal", "mudah digunakan", "tidak membantu"]
labels = [1, 0, 1, 0, 1, 0]
X_tr, X_te, y_tr, y_te = train_test_split(
    texts, labels, test_size=.33, stratify=labels, random_state=42)

model = Pipeline([
    ("tfidf", TfidfVectorizer(ngram_range=(1, 2))),
    ("clf", LogisticRegression(max_iter=1000))
])
model.fit(X_tr, y_tr)
print(classification_report(y_te, model.predict(X_te), zero_division=0))
```
