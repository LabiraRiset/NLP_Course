## Contoh program

**Ringkasan beberapa run**  
Gunakan hasil dari seed yang ditetapkan sebelum eksperimen.


```python
import numpy as np
from scipy import stats

f1_runs = np.array([0.812, 0.821, 0.817, 0.809, 0.824])
mean = f1_runs.mean()
std = f1_runs.std(ddof=1)
margin = stats.t.ppf(.975, len(f1_runs)-1) * std / np.sqrt(len(f1_runs))
print(f"macro-F1 = {mean:.3f} ± {std:.3f}")
print(f"95% CI = [{mean-margin:.3f}, {mean+margin:.3f}]")
```
