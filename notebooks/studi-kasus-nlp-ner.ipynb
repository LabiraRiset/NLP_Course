# Studi Kasus 6 — Named-Entity Recognition

NER memberi label pada span entitas. Pipeline pretrained digunakan untuk demonstrasi inference; fine-tuning memerlukan alignment label kata terhadap subword.


```python
!pip -q install -U transformers
```

```python
from transformers import pipeline
ner=pipeline("token-classification",model="Davlan/xlm-roberta-base-ner-hrl",aggregation_strategy="simple")
texts=["Viny mengajar di Universitas Tarumanagara Jakarta.",
       "Bank Indonesia mengumumkan kebijakan baru pada Senin."]
for text in texts:
    print(text)
    for entity in ner(text):
        print(entity["word"],entity["entity_group"],round(entity["score"],3))
```

Bangun dataset BIO dan evaluasi pada level entitas, bukan hanya token. Kategorikan boundary error, type error, missed entity, dan spurious entity. Domain entity baru memerlukan annotation guideline yang konsisten.
