## Contoh program

**Audit performa per kelompok**  
Group gap hanya sinyal awal; lanjutkan dengan pemeriksaan data dan uncertainty.


```python
import pandas as pd
from sklearn.metrics import f1_score

result = pd.DataFrame({
 "group": ["formal","formal","informal","informal","informal"],
 "y_true": [1,0,1,0,1],
 "y_pred": [1,0,0,0,0]
})

scores = result.groupby("group").apply(
    lambda g: f1_score(g.y_true, g.y_pred, average="macro"),
    include_groups=False
)
print(scores)
print("gap:", scores.max() - scores.min())
```
