## Contoh program

**Mini semantic retrieval**  
Contoh menggunakan TF–IDF agar offline; embedding dapat mengganti vectorizer tanpa mengubah alur.


```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

docs = [
 "Tokenisasi memecah teks menjadi unit.",
 "LSTM menggunakan gate untuk mengatur memori.",
 "Transformer memakai self-attention dan positional encoding."
]
query = "model apa yang menggunakan perhatian pada dirinya sendiri?"
vec = TfidfVectorizer().fit(docs + [query])
D = vec.transform(docs)
q = vec.transform([query])
scores = cosine_similarity(q, D)[0]
for i in scores.argsort()[::-1]:
    print(round(scores[i], 3), docs[i])
```
