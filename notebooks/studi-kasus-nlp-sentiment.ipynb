# Studi Kasus 5 — Analisis Sentimen

Baseline sparse tetap penting sebelum memakai Transformer. Contoh membandingkan word dan character TF–IDF pada pipeline bebas leakage.


```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report,confusion_matrix

data=pd.DataFrame({"text":["layanannya bagus","aplikasi sering error","sangat membantu","tidak berguna",
"baguuus banget","gak recommended","mudah dipakai","mengecewakan sekali"],"label":[1,0,1,0,1,0,1,0]})
Xtr,Xte,ytr,yte=train_test_split(data.text,data.label,test_size=.25,stratify=data.label,random_state=42)
model=Pipeline([("tfidf",TfidfVectorizer(analyzer="char_wb",ngram_range=(3,5))),
                ("clf",LogisticRegression(max_iter=1000))])
model.fit(Xtr,ytr);pred=model.predict(Xte)
print(classification_report(yte,pred,zero_division=0));print(confusion_matrix(yte,pred))
```

Perluas dataset, buat validation set, dan bandingkan word n-gram, character n-gram, serta fine-tuned Transformer. Analisis negasi, ironi, kata tidak baku, dan domain shift.
