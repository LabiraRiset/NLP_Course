## Contoh program

**Similarity pada embedding mini**  
Contoh membangun ruang vektor kecil agar operasi mudah diamati.


```python
import numpy as np

E = {
 "mahasiswa": np.array([.9, .7, .1]),
 "dosen":     np.array([.8, .8, .1]),
 "kampus":    np.array([.7, .6, .4]),
 "mangga":    np.array([.1, .0, .9])
}

def cosine(a, b):
    return float(a @ b / (np.linalg.norm(a) * np.linalg.norm(b)))

query = E["mahasiswa"]
ranking = sorted(
    ((word, cosine(query, vec)) for word, vec in E.items() if word != "mahasiswa"),
    key=lambda x: x[1], reverse=True)
print(ranking)
```
