Bi-encoder menghitung embedding dokumen satu kali. Query baru cukup di-encode lalu dibandingkan dengan index. Bandingkan dengan TF–IDF untuk memahami exact match dan parafrasa.

```python
import numpy as np,pandas as pd
from sentence_transformers import SentenceTransformer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

docs=["Cara melakukan fine-tuning dengan LoRA","Evaluasi retrieval menggunakan Recall dan MRR",
      "Self-attention merupakan komponen utama Transformer","Tokenisasi memecah teks menjadi subword"]
query="bagaimana mengadaptasi model dengan parameter yang sedikit?"

tfidf=TfidfVectorizer().fit(docs+[query]);lex=cosine_similarity(tfidf.transform([query]),tfidf.transform(docs))[0]
model=SentenceTransformer("sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2")
emb=model.encode(docs,normalize_embeddings=True);q=model.encode(query,normalize_embeddings=True);dense=emb@q
pd.DataFrame({"document":docs,"tfidf":lex,"dense":dense}).sort_values("dense",ascending=False)
```

Evaluasi query exact keyword, parafrasa, singkatan, salah eja, campur kode, dan out-of-domain. Laporkan Recall@k, MRR, latency, threshold rejection, serta hard-negative error.
