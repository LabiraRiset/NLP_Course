## Contoh program

**Scaled dot-product attention dari nol**  
Operasi NumPy ini memperlihatkan skor, bobot, dan context.


```python
import numpy as np

def softmax(x):
    z = x - x.max(axis=-1, keepdims=True)
    e = np.exp(z)
    return e / e.sum(axis=-1, keepdims=True)

Q = np.array([[1., 0.]])       # satu query
K = np.array([[1., 0.], [.2, .8], [0., 1.]])
V = np.array([[10., 0.], [4., 6.], [0., 10.]])

scores = Q @ K.T / np.sqrt(K.shape[-1])
weights = softmax(scores)
context = weights @ V
print("scores:", scores)
print("weights:", weights)
print("context:", context)
```
