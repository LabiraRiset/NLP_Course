## Contoh program

**Membandingkan word dan character features**  
Contoh memperlihatkan bahwa pilihan unit menghasilkan vocabulary berbeda.


```python
from sklearn.feature_extraction.text import TfidfVectorizer

texts = [
    "pelayanannya sangat membantu",
    "pelayanan tidak membantu",
    "layanannya membantu banget"
]

word_vec = TfidfVectorizer(ngram_range=(1, 2))
char_vec = TfidfVectorizer(analyzer="char_wb", ngram_range=(3, 5))

X_word = word_vec.fit_transform(texts)
X_char = char_vec.fit_transform(texts)

print("Word features:", X_word.shape)
print(word_vec.get_feature_names_out()[:12])
print("Character features:", X_char.shape)
print(char_vec.get_feature_names_out()[:12])
```
