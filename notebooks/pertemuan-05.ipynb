## Contoh program

**RNN classifier dengan Keras**  
Contoh menggunakan integer sequence sintetis; ganti dengan tokenizer dan dataset kelas.


```python
import numpy as np
import keras
from keras import layers

# 200 dokumen, masing-masing 30 token; 0 dipakai untuk padding
x = np.random.randint(1, 1000, size=(200, 30))
y = (x.mean(axis=1) > 500).astype("int32")

model = keras.Sequential([
    layers.Embedding(input_dim=1000, output_dim=32, mask_zero=True),
    layers.SimpleRNN(32),
    layers.Dense(1, activation="sigmoid")
])
model.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])
model.fit(x, y, validation_split=.2, epochs=5, batch_size=32)
print(model.predict(x[:2]))
```
