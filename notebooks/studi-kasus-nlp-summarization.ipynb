```python
from transformers import pipeline
text="""Natural Language Processing mempelajari cara komputer memproses bahasa manusia. Pendekatan modern menggunakan Transformer dan pretraining. Evaluasi tetap memerlukan baseline, data uji, metrik, dan analisis kesalahan. Untuk sistem generatif, kelancaran tidak menjamin kebenaran fakta."""

# Extractive baseline: dua kalimat pertama
sentences=[x.strip() for x in text.split('.') if x.strip()]
extractive='. '.join(sentences[:2])+'.'

summarizer=pipeline("text2text-generation",model="cahya/mt5-small-indonesian-summarization-cased")
abstractive=summarizer("ringkas: "+text,max_new_tokens=70,do_sample=False)[0]["generated_text"]
print("Extractive:",extractive)
print("Abstractive:",abstractive)
```

Tambahkan reference summary dan hitung ROUGE-1/L, compression ratio, coverage, serta factual error count. Human rubric menilai relevance, coherence, fluency, dan faithfulness.
