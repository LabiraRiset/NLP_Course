## Contoh program

**Prompt benchmark tanpa API tertentu**  
Fungsi call_model sengaja dipisah agar dapat diganti dengan provider atau model lokal.


```python
test_cases = [
    {"input": "Ringkas: NLP membantu komputer memahami teks.",
     "must_contain": "komputer"},
    {"input": "Klasifikasikan sentimen: layanan ini buruk",
     "must_contain": "negatif"}
]

def build_prompt(text):
    return f"""Anda asisten NLP. Jawab singkat dalam Bahasa Indonesia.
INPUT:
{text}
OUTPUT:"""

def evaluate(call_model):
    passed = 0
    for case in test_cases:
        answer = call_model(build_prompt(case["input"]))
        ok = case["must_contain"].lower() in answer.lower()
        passed += ok
        print(ok, answer)
    return passed / len(test_cases)

# score = evaluate(call_model)
```
