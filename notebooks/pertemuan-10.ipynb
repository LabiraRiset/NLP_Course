## Contoh program

**Fine-tuning BERT dengan Transformers**  
Memerlukan internet saat pertama mengunduh model dan GPU disarankan.


```python
# !pip -q install transformers datasets accelerate
from datasets import Dataset
from transformers import AutoTokenizer, AutoModelForSequenceClassification
from transformers import TrainingArguments, Trainer

name = "distilbert-base-multilingual-cased"
tok = AutoTokenizer.from_pretrained(name)
raw = Dataset.from_dict({
    "text": ["pelayanan sangat baik", "aplikasi sering gagal"],
    "label": [1, 0]
})
data = raw.map(lambda b: tok(b["text"], truncation=True), batched=True)
model = AutoModelForSequenceClassification.from_pretrained(name, num_labels=2)
args = TrainingArguments("checkpoints", num_train_epochs=2,
                         per_device_train_batch_size=2, report_to="none")
trainer = Trainer(model=model, args=args, train_dataset=data)
trainer.train()
```
